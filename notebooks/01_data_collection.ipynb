## Project Question

Track A asks whether Shanghai neighborhoods support a healthy 15-minute lifestyle. The working unit is an H3 resolution 8 cell, and the core question is: which cells combine everyday service access with sport, green, fresh-food, cycling-support, and air-quality signals?

## Source Decisions

## Literature Review

This project follows the 15-minute-city idea as a proximity-based planning framework, but treats it as an empirical accessibility question rather than a slogan. Moreno et al. (2021) define the 15-minute city around chrono-urbanism, proximity, density, diversity, and digital support. Their framework is useful for Track A because healthy lifestyle and sport depend on more than one service type: a good cell should combine everyday services, parks, sports facilities, food access, and lower environmental burden. However, the concept is also easy to overstate. A place can be close to many amenities while still being unsafe, unaffordable, polluted, or difficult to use for older residents, children, disabled people, or shift workers. The review therefore separates the planning ideal from the measurement strategy used in this submission.

For measurement methodology, Weng et al. (2019) are the most directly relevant reference because they evaluate 15-minute walkable neighborhoods in Shanghai. Their method adapts a Walk Score style approach, uses amenity categories and pedestrian-group differences, and then tests social inequality in the resulting scores. This is close to the data available here: many POI categories, district information, and a need to make comparable scores across the whole municipality. Their work also shows the weakness of simple amenity counting. Counts must be normalized, grouped by urban need, and interpreted with pedestrian behavior in mind. In this project, baseline access is therefore not a single POI density layer. It combines food, shopping/life services, health, education, transit, and green/public facilities, then scales raw signals into empirical percentiles so that very dense central areas do not automatically dominate every layer.

Logan et al. (2022) generalize this into an x-minute-city measurement problem for 10, 15, and 20 minute thresholds. Their contribution is important because it makes the time threshold itself a modeling choice rather than an obvious truth. A 15-minute threshold can be measured by Euclidean distance, network distance, scheduled transit time, cycling time, or car travel time, and each version answers a different planning question. The brief for this course asks for walk, bike, transit, and car isochrones, with car used for comparison rather than baseline scoring. The present project follows that structure in the interface and score fields, but the current implementation is an H3 neighborhood proxy rather than a full routing model. The limitation is documented because a reproducible proxy is preferable to an unverified claim of true isochrone calculation.

Yang et al. (2023) provide another Shanghai-specific reference by measuring 15-minute community life circles with POI-based walkability. Their work reinforces three practical choices used here. First, the relevant unit should be fine enough to show neighborhood differences, which supports using H3 resolution 8 and a 500m grid proxy. Second, categories need to be balanced because a neighborhood with many restaurants but few health, education, or green facilities should not receive a perfect baseline score. Third, Shanghai must be read as a polycentric and unequal urban region, not only as an inner-city accessibility surface. For this reason, the Web app includes district filtering and not only a global Top10 list.

The Track A layer adds healthy lifestyle and sport. The project operationalizes this with sport and recreation POIs, green/public-space proxies, fresh-food access, bike-support proxies, and district AQI. This is not as strong as the ideal data model in the brief, which also names dedicated cycling lane length, NDVI from satellite imagery, public swimming pools, and more detailed sport facility types. Those indicators are conceptually important because active living depends on the quality, safety, and environmental comfort of the route, not just the presence of destinations. The current dataset can still identify likely sport deserts, but it should be interpreted as a screening layer that needs NDVI, verified cycling infrastructure, and facility subtype validation before being used as policy evidence.

The equity critique matters for interpreting the output. Pereira, Schwanen, and Banister (2017) argue that accessibility should be evaluated as a distributive justice issue, because high average access can hide who is excluded. A city can show strong mean accessibility while peripheral districts, lower-income residents, or people with limited mobility face much weaker practical access. Weng et al. (2019) similarly show that Shanghai walkability advantages concentrate in central areas and differ by population group. This means that a composite score should not be treated as a neutral technical fact. It is a weighted statement about what counts as access, whose needs are prioritized, and which missing data are tolerated.

Pozoukidou and Chatziyiannaki (2021) caution that the 15-minute city is not a completely new planning idea and that implementation must be judged through inclusion, safety, and health, not only proximity. Their critique is especially relevant because walkable, amenity-rich areas often become expensive and may intensify gentrification. Track C in the brief makes this point explicit, but Track A also faces it: healthy lifestyle access may be strongest in wealthy inner districts, while outer districts may have lower service density or worse environmental conditions. The Web app therefore does not present the highest-scoring central H3 cells as a final answer. It exposes district filters, data-source limits, sport-desert gaps, and exportable scores so spatial concentration and tradeoffs can be inspected.

Implementation follows these lessons in a pragmatic way. The notebooks log data provenance, explain why the problematic 2024 education data is replaced by the 2026 Gaode / AMap education POI package, and separate baseline and Track A scoring. The Web app makes the proxy status visible through a data transparency panel. A selected H3 cell is described with strengths, watch points, amenity evidence, AQI, transit proxy, and a rent-band note rather than pretending that every requested indicator has been measured. This transparency is part of the methodology: a useful 15-minute-city dashboard should help users see both accessibility patterns and the uncertainty behind them.

The ideal next step is to replace the H3 neighborhood proxy with network travel-time isochrones for each 500m grid center. That would require a cleaned road graph, mode-specific speeds or transit schedules, and reproducible caching. Until that environment is ready, the current proxy remains useful for comparing broad spatial patterns and communicating where healthy-lifestyle access appears concentrated. It should not be interpreted as precise door-to-door travel time. This distinction is central to the submission: the project demonstrates a documented and reproducible analytical pipeline, while clearly marking which parts are complete, which parts are proxy measures, and which parts require additional verified data.

```python
from pathlib import Path
import csv
import json

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sources = {
    'poi_2023_shp_dir': ROOT / '2023_Shp' / 'Shp',
    'edu_2026_csv': ROOT / 'EDU 2026 POI' / 'preschool_k12_clean.csv',
    'aqi_2025_csv': ROOT / 'shanghai_aqi_official_2025_20250101_20251231_package' / 'clean' / 'shanghai_aqi_district_daily_sthj_official_20250101_20251231.csv',
    'roads_parquet': ROOT / 'shanghai-roads-simplified.parquet',
}
for name, path in sources.items():
    print(name, path.exists(), path)
```

```python
def csv_rows(path):
    with path.open('r', encoding='utf-8-sig', newline='') as f:
        return sum(1 for _ in csv.DictReader(f))

shp_files = sorted(sources['poi_2023_shp_dir'].glob('*.shp'))
print('2023 shapefile groups:', len(shp_files))
print('2026 education POI rows:', csv_rows(sources['edu_2026_csv']))
print('2025 district AQI rows:', csv_rows(sources['aqi_2025_csv']))
print('roads parquet bytes:', sources['roads_parquet'].stat().st_size)
```
